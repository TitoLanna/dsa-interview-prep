## Problem Description
You are given an integer array `height` of length `n`. 

There are `n` vertical lines drawn such that the two endpoints of the `i`th line are **`(i, 0)`** and **`(i, height[i])`**.

Find two lines that together with the `x-axis` form a container, such that the container contains the most water.

Return the maximum amount of water a container can store.

`NB`:Notice that you may not slant the container.


---


 EXAMPLE [1]

````python
import matplotlib.pyplot as plt

height = [1,8,6,2,5,4,8,3,7]

# x-axis positions
x = range(len(height))

# Draw the vertical lines
plt.vlines(x, ymin=0, ymax=height)

# Max container for this example:
# left = 1, right = 8
left = 1
right = 8

container_height = min(height[left], height[right])

# Draw the water area
plt.fill_between(
    [left, right],
    0,
    container_height,
    alpha=0.3
)

# Labels
plt.xlabel("Index")
plt.ylabel("Height")
plt.title("Container With Most Water")

plt.xticks(range(len(height)))

plt.show()

````


```


Input: height = [1,8,6,2,5,4,8,3,7]

Output: 49

Explanation: The above vertical lines are represented by array [1,8,6,2,5,4,8,3,7]. In this case, the max area of water (blue section) the container can contain is 49.
```
---
```
**`Example 2`**

Input: height = [1,1]

Output: 1
```



SOlUTION

````python
def maxArea(arr):
    max_Amount = 0
    left = 0
    right = len(arr)-1
    while left < right:
        
       
       water_level = arr[left] if arr[left]< arr[right] else arr[right]
       
       width_container =right - left 
       
       max_Amount = max(max_Amount,(water_level * width_container))
       if arr[left] < arr[right]:
           left+=1
       else:
           right-=1
    return max_Amount
````

````python
#TEST

test_cases=[
    {
        "input":[1,8,6,2,5,4,8,3,7],
        "output":49
    },
      {
            "input":[1,1],
            "output":1
        }
]
for i,test in enumerate(test_cases):
    exp_out,containers=test["output"],test["input"]
    res =maxArea(containers)
    print(f"Passed [✅] \t {res}" if res ==exp_out else f" Failed! [❌] \t Your result :{res} \t expected: { exp_out}  " )
    
````
