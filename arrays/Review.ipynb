## TWO SUM


```python
#Given a array of integers NUMS and a target .
# Write a algorithm that returns the indices of the two numbers who's sum equals the target

def TwoSum(nums ,target):
    result = {}
    for i in range(len(nums)):
        complement = target -nums[i]
        if complement in result:
            return [result[complement],i]
        result[nums[i]] = i
    return []

#Test
nums,target =[2,7,11,15] ,9
print(TwoSum(nums,target))

""" 
Today I solved this problem without the enumerate function.
 
"""
```

## Best Time to Buy and Sell Stock II

```python
#You are given an integer array prices where prices[i] is the price of a given stock on the [i]th day.
#On each day, you may decide to buy and/or sell the stock. 
#You can only hold at most one share of the stock at any time. 
#However, you can sell and buy the stock multiple times on the same day, 
#ensuring you never hold more than one share of the stock.

#Find and return the maximum profit you can achieve.
""" 
Example 1:

Input: prices = [7,1,5,3,6,4]
Output: 7
Explanation: Buy on day 2 (price = 1) and sell on day 3 (price = 5), profit = 5-1 = 4.
Then buy on day 4 (price = 3) and sell on day 5 (price = 6), profit = 6-3 = 3.
Total profit is 4 + 3 = 7.

Example 2:

Input: prices = [1,2,3,4,5]
Output: 4
Explanation: Buy on day 1 (price = 1) and sell on day 5 (price = 5), profit = 5-1 = 4.
Total profit is 4.
Example 3:

Input: prices = [7,6,4,3,1]
Output: 0
Explanation: There is no way to make a positive profit, so we never buy the stock to achieve the maximum profit of 0.

"""
```

Solution


```python
def max_profit_II(prices):
    
    best_total = 0
    for i in range(len(prices)-1):
      if prices[i]< prices[i+1]:
          best_total+= prices[i+1]-prices[i]    
    return best_total
#Test Cases
test_cases = [
    {
        "input":[7,1,5,3,6,4],"output":7
         },
    {
            "input":[1,2,3,4,5],"output":4
                },
    {
            "input":[7,6,4,3,1],"output":0
                }
]

for i,test in enumerate (test_cases):
    res = max_profit_II(test["input"])
    print(f"Input: {test['input']}","Passed !"if test["output"]==res else "Failed !")
```

## Remove Duplicates from Sorted Array

```python
#Given an integer array nums sorted in non-decreasing order, 
# remove the duplicates in-place such that each unique element appears only once.
# The relative order of the elements should be kept the same.

#Consider the number of unique elements in nums to be k​​​​​​​​​​​​​​. 
# After removing duplicates, return the number of unique elements k.

#The first k elements of nums should contain the unique numbers in sorted order. 
# The remaining elements beyond index k - 1 can be ignored.
"""
Custom Judge:
The judge will test your solution with the following code:

int[] nums = [...]; // Input array
int[] expectedNums = [...]; // The expected answer with correct length

int k = removeDuplicates(nums); // Calls your implementation

assert k == expectedNums.length;
for (int i = 0; i < k; i++) {
    assert nums[i] == expectedNums[i];
}
If all assertions pass, then your solution will be accepted.

 

Example 1:

Input: nums = [1,1,2]
Output: 2, nums = [1,2,_]
Explanation: Your function should return k = 2, with the first two elements of nums being 1 and 2 respectively.
It does not matter what you leave beyond the returned k (hence they are underscores).
Example 2:

Input: nums = [0,0,1,1,1,2,2,3,3,4]
Output: 5, nums = [0,1,2,3,4,_,_,_,_,_]
Explanation: Your function should return k = 5, with the first five elements of nums being 0, 1, 2, 3, and 4 respectively.
It does not matter what you leave beyond the returned k (hence they are underscores).

"""
```

Solution

```python
def remove_duplicate(nums):
    """ 
    Make l the start index as well as the counter of unique elements
    The for loop is used to search for the index of the next unique element
    Whenever we encounter a new element we  place it at the l index and increase it by one
    by doing this we count the number of unique element as well as push the unique elements forward
    the is done in O(n) time 
    """
    l =1
    for r in range(1,len(nums)):
        if nums[r]!= nums[r-1]:
            nums[l]=nums[r]
            l+=1 
    return l

#Test Cases
test_cases =[
    {"input":[1,3,4,4,5,5,6,7,7,8,8,8],"output":7},
    {"input": [0,0,1,1,1,2,2,3,3,4],"output": 5},
    {"input":[1,1,2],"output":2},
    {"input":[0,0,0],"output":1},
    {"input":[1,3],"output":2},
    {"input":[0,1,2],"output":3},
    {"input":[1],"output":1}
    
]
for i,test in enumerate(test_cases):
    res = remove_duplicate(test["input"])
    isPassed = True if res == test["output"] else False
    print(f"test_case [{i+1}]",f"Input:{test['input']}","Passed! "if isPassed else "Failed")
```

## Subarray Sum Equals K

```python
#Given an array of integers nums and an integer k, 
# return the total number of subarrays whose sum equals to k.
# A subarray is a contiguous non-empty sequence of elements within an array.
"""
Example 1:

Input: nums = [1,1,1], k = 2
Output: 2
Example 2:

Input: nums = [1,2,3], k = 3
Output: 2"""
```

solution

```python
def subarraySum(nums , k):
    """ 
    This is the brute force approach, it passed 72/93 test cases but is not the optimum
    """
    ans =0
    freq={}
    for i in range(len(nums)):
        freq[i] = nums[i]
        if freq[i]== k:
            ans+=1
        for j in range(i+1,len(nums)):
            freq[i]+=nums[j]
            if freq[i]==k:
                ans+=1       
    return ans
#Test Cases

test_cases =[
    {"input":[1,1,1],"k":2,"output":2},
    {"input":[1,2,3],"k":3 ,"output":2},
    {"input":[0,3,1],"k":4,"output":2},
    {"input":[0,0,0],"k":0,"output":6}
]
for i,test in enumerate (test_cases):
    res = subarraySum(test["input"],test["k"])
    print(f"test case [{i+1}]: {test['input']} \t","Passed!" if res ==test["output"] else "Failed !")
```
