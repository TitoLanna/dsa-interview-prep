TWO SUM


```python
#Given a array of integers NUMS and a target .
# Write a algorithm that returns the indices of the two numbers who's sum equals the target

def TwoSum(nums ,target):
    result = {}
    for i in range(len(nums)):
        complement = target -nums[i]
        if complement in result:
            return [result[complement],i]
        result[nums[i]] = i
    return []

#Test
nums,target =[2,7,11,15] ,9
print(TwoSum(nums,target))

""" 
Today I solved this problem without the enumerate function.
 
"""
```

Best Time to Buy and Sell Stock

```python
#Given an array of integers PRICES containing stocks  and Indices as days.
#Write a function the returns the maximize profit by choosing a single day to buy stocks,
# And a different day to sell stock
#return 0 if you cannot achieve any profit

def maxProfit(prices):
    current_best_buy = prices[0]
    current_best_profit = 0
    for  i in range(1,len(prices)-1):
        #Get the min of the cheapest stock
        if current_best_buy >= prices[i]:
            current_best_buy = prices[i] 
        current_best_profit =max(current_best_profit,prices[i+1]- current_best_buy)
    
    return current_best_profit
    

print(maxProfit([7,1,5,3,6,4]))#Expected output ->5

"""

"""
```
