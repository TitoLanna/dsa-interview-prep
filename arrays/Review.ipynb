TWO SUM


```python
#Given a array of integers NUMS and a target .
# Write a algorithm that returns the indices of the two numbers who's sum equals the target

def TwoSum(nums ,target):
    result = {}
    for i in range(len(nums)):
        complement = target -nums[i]
        if complement in result:
            return [result[complement],i]
        result[complement] = i

#Test
nums,target =[2,7,11,15] ,9
print(TwoSum(nums,target))
```
