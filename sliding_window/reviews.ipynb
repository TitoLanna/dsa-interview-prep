```python
# Given a string s.
# write a function the returns the length of the longest substring without duplicate characters
```

```python
def longest_substring(s):
    if len(s)<1:
        return 0
    
    left_pointer =0
    max_window_len = 1
    
    seen  = {}
    
    for rigth_pointer,val in enumerate(s):
        
        if val in seen and  seen[val] >=left_pointer:
            left_pointer =seen[val]+1
        seen[val]=rigth_pointer
        cur_window_size = rigth_pointer-left_pointer+1
        max_window_len=max(cur_window_size,max_window_len)
    return max_window_len
    
    
```

Test

```python
test_cases =[
    
    {# Empty string
        "input":"",
        "results":0 
        
    },
     {# string without duplicate
            "input":"asdfghj",
            "results":7 
            
        },
      {# string with 2 elements
             "input":"AB",
             "results":2 
             
         },
       {# string with uppercase and lowercase element
              "input":"abcde",
              "results":5 
              
          }, 
       {# just a basic string
                  "input":"abcddef",
                  "results":4 
                  
        }
    
]


solutions = {}
for i,test in enumerate(test_cases):
    s,exp_res= test["input"],test["results"]
    
    res=longest_substring(s)
    solutions[f"Test_{i+1}"] = [
       f"|Passed !  ||Produced: {res} || Expected : {exp_res}"
       if res == exp_res else 
       f"||Failed !  ||Produced: {res} || Expected : {exp_res}"
    ]
for test in solutions.items():
    print(test)
    
```

# Minimum size Subarray sum
---
- Given a positive integer array and a target.

- write a function that return the minimum length of the sub array whose element sum is equal to the target return 0 otherwise

---
Follow up :

            if you figured out the O(n) solution,try coding another solution of which the time complexity is O(nlogn)


```python
def minSizeSubarray_Sum(arr,target):
    
    
    cur_sum =0
    left = 0
    min_len =float("inf")
    
    for right, val in enumerate(arr):
        cur_sum+=val
        
        while cur_sum >target and left <= right:
            cur_sum-=arr[left]
            left+=1
            
        if cur_sum== target:
            min_len= min(min_len,right-left+1)
    return 0 if min_len == float("inf") else min_len
    
        
    

#TEST CASES

cases =[
    #basic array and target
    {"arr":[2,3,3,4,5,7,1],"target":8,"res":2},
    #No element subarray sum equals the target
    {"arr":[2,4,3,6,7],"target":5,"res":0},
    #Only two elements in the array
    {"arr":[2,8],"target":8,"res":1},
    #One element array
    {"arr":[4],"target":4,"res":1},
    #large array size
    {"arr":[3,2,4,5,1,3,2,3,2,7,8,24,16,8,9,1,5],"target":15,"res":2}
]

solutions ={}
for idx,test in enumerate(cases):
    arr= test["arr"]
    target = test["target"]
    exp_res = test["res"]
    results = minSizeSubarray_Sum(arr,target)
    
    solutions[f"Test_{idx+1}"] = (
      f"Test_[{idx+1}]...Passed !  ||Produced: {results} || Expected : {exp_res}"
           if results == exp_res else 
           f"Test_[{idx+1}]...Failed !  ||Produced: {results} || Expected : {exp_res}"
    )
for test in solutions:
    print(solutions[test])
```
