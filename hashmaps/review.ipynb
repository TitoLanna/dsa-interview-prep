# RandomizedSet — Insert, Delete & GetRandom in O(1)

## Problem Description

Implement the `RandomizedSet` class.

The data structure must support **insertion**, **removal**, and **random element selection** in **average O(1) time complexity**.

---

## Required Methods

### `RandomizedSet()`

Initializes a new `RandomizedSet` object.

### `insert(int val)`

Inserts `val` into the set **only if it is not already present**.

**Returns:**

- `True` if `val` was successfully inserted.
- `False` if `val` was already present.

**Target Time Complexity:** `O(1)` average

### `remove(int val)`

Removes `val` from the set **if it is present**.

**Returns:**

- `True` if `val` was successfully removed.
- `False` if `val` was not present.

**Target Time Complexity:** `O(1)` average

### `getRandom()`

Returns a random element from the current set.

It is guaranteed that at least one element exists when this method is called.

Each element must have an **equal probability** of being selected.

**Target Time Complexity:** `O(1)`

---

## Example

### Input

```text
["RandomizedSet", "insert", "remove", "insert", "getRandom", "remove", "insert", "getRandom"]

[[], [1], [2], [2], [], [1], [2], []]
```

### Output

```text
[null, true, false, true, 2, true, false, 2]
```

> **Note:** The first call to `getRandom()` may return either `1` or `2` because both elements are present at that point.

---

## Step-by-Step Explanation

```python
randomizedSet = RandomizedSet()

randomizedSet.insert(1)
# Returns True
# Set now contains: [1]

randomizedSet.remove(2)
# Returns False
# 2 does not exist in the set.

randomizedSet.insert(2)
# Returns True
# Set now contains: [1, 2]

randomizedSet.getRandom()
# Returns either 1 or 2 randomly.

randomizedSet.remove(1)
# Returns True
# Set now contains: [2]

randomizedSet.insert(2)
# Returns False
# 2 is already present.

randomizedSet.getRandom()
# Returns 2
# 2 is now the only element in the set.
```

## Complexity Requirements

| Operation | Required Average Time Complexity |
|---|---:|
| `insert(val)` | **O(1)** |
| `remove(val)` | **O(1)** |
| `getRandom()` | **O(1)** |
| Space Complexity | **O(n)** |

---

## Key Challenge

A normal Python `set` provides average **O(1)** insertion and removal:

```python
seen.add(val)
seen.remove(val)
```

However, selecting a random element using:

```python
random.choice(list(seen))
```

is **O(n)** because converting the set into a list requires visiting all `n` elements.

The challenge is therefore:

> **How can we combine data structures so that insertion, removal, and random selection are all O(1) on average?**

### Hint

Consider combining:

- A **list** for O(1) random access.
- A **hash map / dictionary** for O(1) lookup.
- A **swap-with-last-and-pop** technique for O(1) removal.

````python

import random
class RandomizedSet(object):
    
    def __init__(self):
        self. nums =[]
        self. idx_map={}
        
    def search(self,val):
        return val in self.nums
    
    def insert(self,val):
        if self.search(val):
            return False
        self.nums.append(val)
        self.idx_map[val]= len(self.nums)-1
        return True
    
    def remove(self,val):
        if not self.search(val):
            return False
        
        index = self.idx_map[val]
        last_element = self.nums[-1]
        
        self.nums[index]=last_element
        self.nums.pop()
        
        self.idx_map[last_element] = index
        del self.idx_map[val]
        return True
    
    def getRandom(self):
       return random.choice(self.nums)
    
# --------------------------
# TEST
# --------------------------

obj = RandomizedSet()

print(obj.insert(1))     # True
print(obj.remove(2))     # False
print(obj.insert(2))     # True
print(obj.getRandom())   # 1 or 2
print(obj.remove(1))     # True
print(obj.insert(2))     # False
print(obj.getRandom())   # 2
````
